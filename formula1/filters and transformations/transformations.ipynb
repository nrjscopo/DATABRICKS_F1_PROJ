```python
circuits_df = spark.read.parquet(f"{processed_folder_path}/circuits")
```

```python
races_df = spark.read.parquet(f"{processed_folder_path}/races").filter("year=2019")
```

```python
races_rena_df = races_df.withColumnRenamed("name", "racesName")
```

```python
circuits_rena_df = circuits_df.withColumnRenamed("circuit_id", "circuitId") \
    .withColumnRenamed("name", "circuitsName")
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
race_circuits_df = circuits_rena_df.join(races_rena_df, "circuitId", "inner").select(circuits_rena_df.circuitsName, circuits_rena_df.location, circuits_rena_df.country, races_rena_df.racesName, races_rena_df.round).display()
```
